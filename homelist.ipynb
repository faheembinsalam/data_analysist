```python
# Given list of subject marks out of 100 each
subject_marks = [80, 50, 60, 75, 85]

# Total possible marks (5 subjects * 100)
total_marks = 500

# Step 1: Sum all values using a for loop
obtained_marks = 0
for mark in subject_marks:
    obtained_marks += mark  # Adding each mark to obtained_marks

# Step 2: Calculate Percentage
percentage = (obtained_marks / total_marks) * 100

# Displaying the results
print(f"Obtained Marks: {obtained_marks} / {total_marks}")
print(f"Percentage: {percentage:.2f}%")
```

```python
numbers = [100, 99, 2, 0, 9, -9, 0, 11, 0, 1000]
zero_count = 0

for num in numbers:
    if num == 0:
        zero_count += 1

print("Total number of zeros:", zero_count)
```
