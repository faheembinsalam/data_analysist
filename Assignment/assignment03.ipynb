**# 1. Write a program that accepts a list from user and print the alternate element of list.**

```python
lst = []

n = int(input("Enter number of elements: "))

for i in range(n):
    item = input("Enter element: ")
    lst.append(item)

print("Alternate elements of list are:")

for i in range(0, len(lst), 2):
    print(lst[i])
```

**#2. Write a program that accepts a list from user. Your program should reverse the content of list and display it. Do not use reverse() method.**

```python
lst=[]
n=int(input("Enter number of elements:"))
for i in range(n):
    item=input("Enter element: ")
    lst.append(item)

rev_list = []
for i in range(len(lst)-1, -1, -1):
    rev_list.append(lst[i])

print("Original list:", lst)
print("Reversed list:", rev_list)
```

**#3. Find and display the largest number of a list without using built-in function max(). Your program should ask the user to input values in list from keyboard.**

```python
lst = []
n = int(input("Enter number of elements: "))

for i in range(n):
    num = int(input("Enter number: "))
    lst.append(num)
largest = lst[0]

for i in range(1, len(lst)):
    if lst[i] > largest:
        largest = lst[i]
print("List:", lst)
print("Largest number is:", largest)
```

**#4. Write a program that rotates the element of a list so that the element at the first index moves to the second index, the element in the second index moves to the third index, etc., and the element in the last index moves to the first index.**

```python
lst = []
n = int(input("Enter number of elements: "))

for i in range(n):
    item = input("Enter element: ")
    lst.append(item)

rotated_lst = []
rotated_lst.append(lst[-1])

for i in range(0, len(lst)-1):
    rotated_lst.append(lst[i])

print("Original list:", lst)
print("Rotated list:", rotated_lst)
```

**#6. Write a program that reads a string from the user containing a date in the form mm/dd/yyyy. It should print the date in the form March 12, 2021.**

**#7. Write a program with a function that accepts a string from keyboard and create a new string after converting character of each word capitalized. For instance, if the sentence is "stop and smell the roses."
the output should be "Stop And Smell The Roses"**

```python
def capitalized(sentence):
    words = sentence.split()
    new_string = ""

    for word in words:
        new_string += word[0].upper() + word[1:] + " "

    return new_string

sentence = input("Enter a sentence: ")
answer = capitalized(sentence)
print(answer)
```

**#8. Find the sum of each row of matrix of size m x n. For example for the following matrix output will be
like this :
Sum of row 1 = 32
Sum of row 2 = 31
Sum of row 3 = 63**

```python
m = int(input("Enter number of rows: "))
n = int(input("Enter number of columns: "))

matrix = []

for i in range(m):
    row = []
    for j in range(n):
        value = int(input("Enter element: "))
        row.append(value)
    matrix.append(row)

for i in range(m):
    total = 0

    for j in range(n):
        total = total + matrix[i][j]

    print("Sum of row", i + 1, "=", total)
```

**#9. Write a program to add two matrices of size n x m.**

```python
n = int(input("Enter number of rows: "))
m = int(input("Enter number of columns: "))

matrix1 = []
print("Enter first matrix:")

for i in range(n):
    row = []
    for j in range(m):
        row.append(int(input("Enter element: ")))
    matrix1.append(row)

matrix2 = []
print("Enter second matrix:")

for i in range(n):
    row = []
    for j in range(m):
        row.append(int(input("Enter element: ")))
    matrix2.append(row)

result = []

for i in range(n):
    row = []
    for j in range(m):
        row.append(matrix1[i][j] + matrix2[i][j])
    result.append(row)
print("Sum of two matrices:")

for row in result:
    print(row)
```

**#10. Write a program to multiply two matrices**

```python
r1 = int(input("Enter rows of first matrix: "))
c1 = int(input("Enter columns of first matrix: "))

matrix1 = []
print("Enter first matrix:")

for i in range(r1):
    row = []
    for j in range(c1):
        row.append(int(input("Enter element: ")))
    matrix1.append(row)

r2 = int(input("Enter rows of second matrix: "))
c2 = int(input("Enter columns of second matrix: "))

if c1 != r2:
    print("Matrix multiplication is not possible")

else:
    matrix2 = []
    print("Enter second matrix:")

    for i in range(r2):
        row = []
        for j in range(c2):
            row.append(int(input("Enter element: ")))
        matrix2.append(row)

    result = []

    for i in range(r1):
        row = []
        for j in range(c2):
            total = 0

            for k in range(c1):
                total = total + matrix1[i][k] * matrix2[k][j]

            row.append(total)

        result.append(row)
        
    print("Multiplication of two matrices:")

    for row in result:
        print(row)
```
