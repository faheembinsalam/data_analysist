```python
# 3. Write a program in which use all the operators we can use in Python

# Student Information
student_name = input("Enter Student Name: ")

math_marks = float(input("Enter Math Marks: "))
english_marks = float(input("Enter English Marks: "))
computer_marks = float(input("Enter Computer Marks: "))

total_obtained = 0.0
total_obtained += math_marks
total_obtained += english_marks
total_obtained += computer_marks

total_max_marks = 300.0
percentage = (total_obtained / total_max_marks) * 100

is_valid_data = True
if percentage is not None:
    grade = None

    if percentage >= 80:
        grade = "A+"
    elif percentage >= 70 and percentage < 80:
        grade = "A"
    elif percentage >= 60 and percentage < 70:
        grade = "B"
    elif percentage >= 50 and percentage < 60:
        grade = "C"
    else:
        grade = "F (Fail)"

    has_failed_subject = (math_marks < 33) or (english_marks < 33) or (computer_marks < 33)

    is_passed = (grade != "F (Fail)") and (has_failed_subject is False)

    print(f"Math: {math_marks}/100 | English: {english_marks}/100 | Computer: {computer_marks}/100")
    print(f"Total Marks Obtained: {total_obtained} / {total_max_marks}")
    print(f"Percentage: {percentage:.2f}%")
    print(f"Grade: {grade}")

    if is_passed is True:
        print("Final Status: PASSED 🎉")
    else:
        print("Final Status: FAILED ❌")
    print("------------------------------------")
```

```python
# 6. Take the radius of a circle from a variable and calculate its Area and Circumference (use pi = 3.14159)

radius = float(input("Enter the radius of the circle: "))
pi = 3.14159

area = pi * (radius ** 2)
circumference = 2 * pi * radius

print(f"Radius Entered: {radius}")
print(f"Area of Circle: {area:.2f}")
print(f"Circumference of Circle: {circumference:.2f}")
```

```python
# 7. Take the price of an item and a discount percentage in variables, then calculate and print the discount amount and the final price after discount

original_price = float(input("Enter the original price of the item ($): "))
discount_percent = float(input("Enter the discount percentage (%): "))

discount_amount = (original_price * discount_percent) / 100
final_price = original_price - discount_amount

print(f"Original Price:   ${original_price:.2f}")
print(f"Discount ({discount_percent}%):  -${discount_amount:.2f}")
print(f"Final Price:      ${final_price:.2f}")
```
